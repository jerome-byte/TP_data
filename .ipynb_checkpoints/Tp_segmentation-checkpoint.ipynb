## 3. Nettoyage des transactions (Partie 1 du TP)

Règles de nettoyage, chacune justifiée et quantifiée :
- retirer les doublons exacts (erreurs d'import/export du fichier source)
- retirer les lignes dont le StockCode n'est pas un vrai produit (POST, D, M, BANK CHARGES...)
- retirer les annulations (Invoice commence par « C »)
- retirer les lignes sans `Customer ID` (achat non attribuable à un client)
- retirer les quantités ou prix négatifs/nuls (erreurs de saisie ou ajustements comptables)

On garde aussi une **deuxième version** avec les annulations nettes,
utilisée plus tard pour mesurer leur impact sur les segments
(question du rapport : « le traitement des annulations change-t-il les segments ? »).

```python
nb_lignes_brut = len(df_brut)

# On enlève d'abord les doublons exacts et les StockCode non-produits,
# avant même de traiter les annulations.
df_brut = df_brut.drop_duplicates()
df_brut = df_brut[df_brut["StockCode"].astype(str).str.match(r"^\d")]
df_brut["est_annulation"] = df_brut["Invoice"].astype(str).str.startswith("C")

print(f"Après retrait doublons + StockCode non-produits : {len(df_brut)} lignes")
```

```python
# Version principale : nettoyage strict
df_clean = df_brut[
    (~df_brut["est_annulation"])
    & (df_brut["Quantity"] > 0)
    & (df_brut["Price"] > 0)
    & (df_brut["Customer ID"].notna())
].copy()
df_clean["Montant"] = df_clean["Quantity"] * df_clean["Price"]

pct_conserve = len(df_clean) / nb_lignes_brut * 100
print(f"Lignes conservées : {len(df_clean)} / {nb_lignes_brut} ({pct_conserve:.1f} %)")
print("Clients restants :", df_clean["Customer ID"].nunique())
```

```python
# Version alternative : on garde les annulations nettes (comparaison plus tard)
df_avec_annulations = df_brut[
    (df_brut["Quantity"] != 0)
    & (df_brut["Price"] > 0)
    & (df_brut["Customer ID"].notna())
].copy()
df_avec_annulations["Montant"] = df_avec_annulations["Quantity"] * df_avec_annulations["Price"]
print("Lignes (version avec annulations nettes) :", len(df_avec_annulations))
```

```python
# Vérification visuelle : les distributions sont plus propres après nettoyage
figure, axes = plt.subplots(1, 2, figsize=(11, 4))
sns.histplot(df_clean["Quantity"], bins=50, ax=axes[0], log_scale=(True, False))
axes[0].set_title("Quantity (nettoyée, échelle log)")
sns.histplot(df_clean["Price"], bins=50, ax=axes[1], log_scale=(True, False))
axes[1].set_title("Price (nettoyée, échelle log)")
plt.tight_layout()
plt.show()
```

## 4. Construction des variables RFM (Partie 2 du TP)

Pour chaque client :
- **Récence** = nombre de jours depuis son dernier achat
- **Fréquence** = nombre de factures distinctes
- **Montant** = somme totale dépensée

```python
def calculer_rfm(transactions):
    """Calcule Récence / Fréquence / Montant par client."""
    date_reference = transactions["InvoiceDate"].max() + pd.Timedelta(days=1)
    rfm = transactions.groupby("Customer ID").agg(
        Recence=("InvoiceDate", lambda dates: (date_reference - dates.max()).days),
        Frequence=("Invoice", "nunique"),
        Montant=("Montant", "sum"),
    ).reset_index()
    return rfm

rfm = calculer_rfm(df_clean)
print("Nombre de clients :", len(rfm))
rfm.describe()
```

```python
# Les 3 variables sont très asymétriques (quelques gros clients, beaucoup de petits)
print("Asymétrie (skew) avant transformation :")
print(rfm[["Recence", "Frequence", "Montant"]].skew())

figure, axes = plt.subplots(1, 3, figsize=(14, 3.5))
sns.histplot(rfm["Recence"], bins=40, ax=axes[0]); axes[0].set_title("Récence")
sns.histplot(rfm["Frequence"], bins=40, ax=axes[1]); axes[1].set_title("Fréquence")
sns.histplot(rfm["Montant"], bins=40, ax=axes[2]); axes[2].set_title("Montant")
plt.tight_layout()
plt.show()
```

## commentaire sur l'asymetrie
`la grande majorite des clients achetent peu et depensent de petite sommes,
Tantdis que nous avons une petite minorite qui achete tres souvent et depense beaucoup de somme`

## 5. Normalisation
k-means calcul des distances geometrique entre les clients
si le montant varie de 0 a 600 000 et la frequence  de 1 a 400. la varisble Montant aura une influence 
Mathematique tellement enorme  qui peut ecraser les autre ce qui va faire que Algo ne verra pas les autre var
pour eviter cela passons par  Deux étapes :
- **log(1+x)** pour réduire l'asymétrie (beaucoup de petites valeurs, quelques très grandes)
- **StandardScaler** pour ramener chaque variable à une échelle comparable (moyenne 0, écart-type 1)

```python
rfm["Recence_log"] = np.log1p(rfm["Recence"])
rfm["Frequence_log"] = np.log1p(rfm["Frequence"])
rfm["Montant_log"] = np.log1p(rfm["Montant"])

colonnes_rfm = ["Recence_log", "Frequence_log", "Montant_log"]
print("Asymétrie après log :")
print(rfm[colonnes_rfm].skew())

scaler = StandardScaler()
X = scaler.fit_transform(rfm[colonnes_rfm])
print("\nMoyenne et écart-type après standardisation (doivent être ~0 et ~1) :")
print(pd.DataFrame(X, columns=colonnes_rfm).agg(["mean", "std"]))
```

```python
# Fréquence et Montant sont naturellement corrélés : un client qui achète
# souvent dépense mécaniquement plus. On mesure cette redondance.
correlation = rfm[colonnes_rfm].corr()
print("Corrélation Fréquence / Montant :", round(correlation.loc["Frequence_log", "Montant_log"], 3))

sns.heatmap(correlation, annot=True, cmap="coolwarm", center=0)
plt.title("Corrélation entre variables RFM (log)")
plt.show()
```

## explication 

- frequence et montant: nous avons une forte correlation positive ce qui signifie que les clints qui commande le plus souvent sont aussi ceux qui depenses le plus au total 
- Recences et autre : correlation negative qui indique que les clients dont de dernier achats remonte a longtemps ont une frequence d'achats et montants faibles, ces clients inatif achetent moins. 

**Traitement de la redondance (question 3 du rapport)** : une corrélation
forte entre Fréquence et Montant signifie qu'ils portent en partie la
même information le clustering risque de sur-pondérer implicitement
cette dimension "engagement" par rapport à la Récence. Deux options sont:
1. garder les 3 variables telles quelles (option par défaut, plus simple à interpréter)
2. réduire Fréquence + Montant à une seule composante par ACP, combinée à la Récence

On compare les deux par silhouette pour confirmer avec un chiffre, pas seulement une intuition.

```python
# Option 2 : ACP sur Fréquence + Montant (très corrélés) en une seule
# composante "engagement", combinée à la Récence gardée telle quelle.
pca_fm = PCA(n_components=1)
engagement = pca_fm.fit_transform(rfm[["Frequence_log", "Montant_log"]])
X_alternatif = np.column_stack([
    StandardScaler().fit_transform(rfm[["Recence_log"]]),
    StandardScaler().fit_transform(engagement),
])

labels_defaut = KMeans(n_clusters=4, random_state=GRAINE, n_init=10).fit_predict(X)
labels_alternatif = KMeans(n_clusters=4, random_state=GRAINE, n_init=10).fit_predict(X_alternatif)

print("Silhouette avec 3 variables (défaut)      :", round(silhouette_score(X, labels_defaut), 3))
print("Silhouette avec ACP Fréq+Montant (2 axes) :", round(silhouette_score(X_alternatif, labels_alternatif), 3))
print("Accord entre les deux approches (ARI)     :", round(adjusted_rand_score(labels_defaut, labels_alternatif), 3))
```

**Décision** : La silhouette de l'option ACP (0.396) est légèrement supérieure à celle des 3 variables (0.365), mais l'écart est faible et l'Adjusted Rand Index entre les deux approches est élevé (0.784) : les segments obtenus sont très similaires. On conserve donc l'option par défaut (3 variables), plus simple à expliquer à un public marketing ("Récence, Fréquence, Montant" est bien plus parlant qu'un axe d'ACP). Le gain statistique marginal ne justifie pas la perte d'interprétabilité.

## 6. Baseline : segmentation par règles (sans machine learning)

Avant de faire du clustering, on construit une **baseline simple** :
un score RFM classique, où chaque variable est découpée en quartiles
(1 = moins bon, 4 = meilleur), puis additionnée. C'est la méthode
utilisée depuis longtemps en marketing, sans aucun algorithme.

Cette baseline sert de point de comparaison : le clustering K-means
doit apporter quelque chose en plus (segments mieux séparés, plus
homogènes) pour être justifié.

```python
# Score de 1 (moins bon) à 4 (meilleur) sur chaque variable.
# Pour la Récence, un client "meilleur" est un client récent, donc on inverse.
rfm["score_R"] = pd.qcut(rfm["Recence"], 4, labels=[4, 3, 2, 1]).astype(int)
rfm["score_F"] = pd.qcut(rfm["Frequence"].rank(method="first"), 4, labels=[1, 2, 3, 4]).astype(int)
rfm["score_M"] = pd.qcut(rfm["Montant"], 4, labels=[1, 2, 3, 4]).astype(int)
rfm["score_RFM"] = rfm["score_R"] + rfm["score_F"] + rfm["score_M"]

def segment_baseline(score):
    if score >= 10:
        return "Meilleurs clients"
    elif score >= 7:
        return "Clients réguliers"
    elif score >= 5:
        return "Clients occasionnels"
    else:
        return "Clients à relancer"

rfm["segment_baseline"] = rfm["score_RFM"].apply(segment_baseline)
print(rfm["segment_baseline"].value_counts())
```

```python
# Limite de cette baseline : les seuils (10, 7, 5) sont fixés à la main,
# regrouper les client par categorie
resume_baseline = rfm.groupby("segment_baseline").agg(
    Effectif=("Customer ID", "count"),
    Montant_moyen=("Montant", "mean"),
).round(0)
resume_baseline
```

## explication 
- clients a relancer (1147): ont un montant moyen tres bas
- client occasionnels : depenses en moyenne
- client reguliers : representent le groupe le plus nombreux avec un panier moyen de 13940
- client vip : un groupe tres forte avec un montant eleve de 8244
## limites de cette methode
- les seuil de score sont fixe a la main d'une maniere arbitraire
- cette division humain ne repose pas mathematiquement sur la structure reelle des donnees

## 7. Modélisation : clustering K-means (Partie 3 du TP)

on combine trois indicateurs pour trouver la meilleur k

```python
## Méthode du coude : on regarde à partir de quel k l'inertie
## (dispersion à l'intérieur des clusters) diminue beaucoup moins vite.
valeurs_k = range(2, 9)
inerties = []
for k in valeurs_k:
    modele = KMeans(n_clusters=k, random_state=GRAINE, n_init=10)
    modele.fit(X)
    inerties.append(modele.inertia_)

plt.plot(list(valeurs_k), inerties, marker="o")
plt.xlabel("Nombre de clusters (k)")#Plus le nombre de groupes augmente, plus l'inertie diminue (ce qui est normal, les groupes étant plus petits, les points sont plus proches de leur centre).On cherche l'endroit où la courbe forme une cassure en forme de coude.Sur notre graphique, ce coude se situe  à k = 4 : à partir de 4 clusters, la courbe commence à s'aplatir fortement, ce qui signifie que rajouter des groupes supplémentaires n'apporte plus d'amélioration significative. C'est donc la valeur de k qu'il est retenue pour la suite 
plt.ylabel("Inertie")
plt.title("Méthode du coude")
plt.show()
```

```python
# Score de silhouette : mesure si les clusters sont bien séparés
# (proche de 1 = bien séparés, proche de 0 = clusters qui se chevauchent).
 
scores_silhouette = []
for k in valeurs_k:
    labels = KMeans(n_clusters=k, random_state=GRAINE, n_init=10).fit_predict(X)
    scores_silhouette.append(silhouette_score(X, labels))

plt.plot(list(valeurs_k), scores_silhouette, marker="o", color="darkorange")
plt.xlabel("Nombre de clusters (k)")
plt.ylabel("Score de silhouette")
plt.title("Score de silhouette par k")
plt.show()

pd.DataFrame({"k": list(valeurs_k), "inertie": inerties, "silhouette": scores_silhouette})
```

**Lecture du tableau** : la silhouette la plus haute
est souvent obtenue à k=2 (un simple « actif » vs « inactifs »). Ce
n'est pas une erreur de méthode, mais k=2 est trop pauvre pour piloter
des actions marketing différenciées. On regarde donc, en plus de la
silhouette, la stabilité (cellule suivante) et le nombre de segments
réellement utile pour le métier, en gardant k=2 comme référence de
comparaison plutôt qu'en l'ignorant.

```python
# Stabilité : un bon k doit donner des clusters similaires même en
# ré-échantillonnant les données (sinon le découpage est un artefact du hasard).
# On compare, pour chaque k, le clustering sur 100% des données vs un
# tirage à 80% des données, avec le score ARI (1 = identique, 0 = aléatoire).
def mesurer_stabilite(X, k, nb_essais=10):
    labels_complet = KMeans(n_clusters=k, random_state=GRAINE, n_init=10).fit_predict(X)
    scores_ari = []
    for graine in range(nb_essais):
        rng = np.random.RandomState(graine)
        echantillon = rng.choice(len(X), size=int(0.8 * len(X)), replace=False)
        labels_echantillon = KMeans(n_clusters=k, random_state=graine, n_init=10).fit_predict(X[echantillon])
        scores_ari.append(adjusted_rand_score(labels_complet[echantillon], labels_echantillon))
    return np.mean(scores_ari)

stabilite_par_k = {k: mesurer_stabilite(X, k) for k in [2, 3, 4, 5, 6]}
print("Stabilité (ARI moyen) par k :", {k: round(v, 3) for k, v in stabilite_par_k.items()})
```

**Décision finale sur k** : k=2 a la silhouette la plus haute mais ne
distingue que deux groupes triviaux, peu actionnables. On choisit donc
k=4, en s'appuyant sur : une silhouette raisonnable (pas la meilleure,
mais correcte), une stabilité élevée (proche de celle de k=2 et k=3), et
surtout un nombre de segments qui correspond à une lecture marketing
standard (Champions / Réguliers / À risque / Endormis). Ce choix est un
compromis assumé entre performance statistique pure et utilité métier —
à documenter explicitement ainsi dans le rapport plutôt que de présenter
k=4 comme le optimum statistique absolu.

```python
K = 4

# n_init=50 : plus de tirages aléatoires initiaux que la valeur par défaut,
# pour réduire le risque de tomber sur un minimum local instable.
kmeans = KMeans(n_clusters=K, random_state=GRAINE, n_init=50)
rfm["cluster"] = kmeans.fit_predict(X)
print(f"k retenu : {K}")
print(rfm["cluster"].value_counts().sort_index())
```

Affiche le nombre de clients répartis dans chacun des 4 groupes finaux.

```python
# Visualisation des clusters en 2D (réduction par ACP pour pouvoir les voir)
pca = PCA(n_components=2)
coords = pca.fit_transform(X)
rfm["axe1"], rfm["axe2"] = coords[:, 0], coords[:, 1]

plt.figure(figsize=(7, 5))
sns.scatterplot(data=rfm, x="axe1", y="axe2", hue="cluster", palette="tab10", alpha=0.6)
plt.title(f"Clusters K-means (k={K}), projection ACP")
plt.show()
```

visualisation en 2D des clusters K-means grâce à une technique de réduction de dimension, l'ACP (Analyse en Composantes Principales).

Comme les données RFM comportent plusieurs dimensions, l'ACP permet de les "aplatir" pour les observer sous forme de carte à plat :

Les couleurs (les 4 clusters) : On distingue nettement 4 grands groupes de couleurs bien distincts (bleu, orange, vert, rouge). Cela prouve que l'algorithme K-means a réussi à partitionner l'espace client de manière cohérente et géométriquement séparée.

## 8. Évaluation : K-means vs baseline

On compare le clustering K-means à la baseline par règles, pour vérifier
que le clustering apporte réellement quelque chose.

```python
silhouette_kmeans = silhouette_score(X, rfm["cluster"])
print("Score de silhouette K-means (k retenu) :", round(silhouette_kmeans, 3))

# Croisement des deux segmentations : est-ce que les clients "Meilleurs clients"
# de la baseline se retrouvent surtout dans un seul cluster K-means ?
tableau_croise = pd.crosstab(rfm["segment_baseline"], rfm["cluster"])
tableau_croise
```

**Lecture du tableau croisé** : si chaque ligne de la baseline se
concentre surtout dans une ou deux colonnes K-means, les deux approches
sont cohérentes entre elles ce qui renforce la confiance dans les
segments trouvés, même sans vérité terrain pour les valider formellement.

## 9. Analyse des erreurs et sensibilité

Sans vérité terrain, on ne peut pas parler d'« erreurs » comme en
apprentissage supervisé. On regarde plutôt la **robustesse** du
clustering : est-ce que le résultat change beaucoup si on modifie
légèrement les données ou le nettoyage ?

```python
# Sensibilité aux valeurs extrêmes : on retire le 1% de clients avec le
# montant le plus élevé, et on refait le clustering pour voir si les
# segments changent beaucoup (mesuré par ARI sur les clients communs).
seuil_extreme = rfm["Montant"].quantile(0.99)
rfm_sans_extremes = rfm[rfm["Montant"] <= seuil_extreme].copy()
X_sans_extremes = scaler.fit_transform(rfm_sans_extremes[colonnes_rfm])
labels_sans_extremes = KMeans(n_clusters=K, random_state=GRAINE, n_init=10).fit_predict(X_sans_extremes)

clients_communs = rfm_sans_extremes["Customer ID"].values
labels_originaux_sur_communs = rfm.set_index("Customer ID").loc[clients_communs, "cluster"].values
ari_extremes = adjusted_rand_score(labels_originaux_sur_communs, labels_sans_extremes)
print(f"Clients extrêmes retirés : {len(rfm) - len(rfm_sans_extremes)}")
print(f"ARI (avec vs sans les 1% plus gros clients) : {ari_extremes:.3f}")
```

```python
# Impact du traitement des annulations (question du rapport, Partie 1)
rfm_avec_annulations = calculer_rfm(df_avec_annulations)
for col, col_log in [("Recence", "Recence_log"), ("Frequence", "Frequence_log"), ("Montant", "Montant_log")]:
    # clip(lower=0) : avec les annulations nettes, le Montant total d'un client
    # peut être négatif (plus de retours que d'achats sur la période).
    # log1p n'est défini que pour x > -1, donc on ramène les valeurs négatives
    # à 0 avant transformation, sinon KMeans reçoit des NaN et refuse de s'exécuter.
    rfm_avec_annulations[col_log] = np.log1p(rfm_avec_annulations[col].clip(lower=0))

X_avec_annulations = StandardScaler().fit_transform(rfm_avec_annulations[colonnes_rfm])
labels_avec_annulations = KMeans(n_clusters=K, random_state=GRAINE, n_init=10).fit_predict(X_avec_annulations)

clients_communs2 = list(set(rfm["Customer ID"]) & set(rfm_avec_annulations["Customer ID"]))
a = rfm.set_index("Customer ID").loc[clients_communs2, "cluster"]
b = pd.Series(labels_avec_annulations, index=rfm_avec_annulations["Customer ID"]).loc[clients_communs2]
ari_annulations = adjusted_rand_score(a.values, b.values)
print(f"ARI (sans annulations vs avec annulations nettes) : {ari_annulations:.3f}")
```

**Grille de lecture pour interpréter l'ARI** (à utiliser aussi pour les
autres tests de robustesse du notebook) :
- ARI > 0,90 : accord très fort, impact négligeable
- ARI entre 0,70 et 0,90 : accord modéré, une part notable de clients change de segment à signaler explicitement dans le rapport, sans remettre en cause le clustering
- ARI < 0,70 : accord faible, le choix de traitement (ici : garder ou non les annulations) change substantiellement les segments et doit être justifié comme un choix de méthode, pas comme un détail

Appliquez cette grille à la valeur obtenue ci-dessus sur vos vraies
données pour rédiger une phrase de conclusion dans le rapport (par
exemple : « le traitement des annulations a un impact modéré, environ
X % des clients changent de segment, principalement en bordure entre
Réguliers et À risque »).

## 10. Caractérisation des segments (Partie 4 du TP)

```python
tableau_segments = rfm.groupby("cluster").agg(
    Effectif=("Customer ID", "count"),
    Recence_moyenne=("Recence", "mean"),
    Frequence_moyenne=("Frequence", "mean"),
    Montant_moyen=("Montant", "mean"),
    CA_total=("Montant", "sum"),
).reset_index()
tableau_segments["pct_CA"] = (tableau_segments["CA_total"] / tableau_segments["CA_total"].sum() * 100).round(1)
tableau_segments = tableau_segments.round(1).sort_values("Montant_moyen", ascending=False)
tableau_segments
```

```python
# Pays et produit dominants par segment, sur la base du CA (Montant),
# pas du nombre de lignes : sinon on ne trouve que le pays/produit le plus
# fréquent, pas celui qui rapporte le plus.
rfm_avec_pays = df_clean.merge(rfm[["Customer ID", "cluster"]], on="Customer ID")

ca_par_cluster_pays = rfm_avec_pays.groupby(["cluster", "Country"])["Montant"].sum()
top_pays = ca_par_cluster_pays.groupby("cluster").idxmax().apply(lambda x: x[1])

ca_par_cluster_produit = rfm_avec_pays.groupby(["cluster", "Description"])["Montant"].sum()
top_produit = ca_par_cluster_produit.groupby("cluster").idxmax().apply(lambda x: x[1])

tableau_segments["pays_dominant"] = tableau_segments["cluster"].map(top_pays)
tableau_segments["produit_dominant"] = tableau_segments["cluster"].map(top_produit)
tableau_segments
```

```python
# Le Royaume-Uni domine presque toujours (marché principal de l'entreprise).
# On regarde aussi le meilleur pays hors UK par segment, plus informatif
# pour d'éventuelles actions d'expansion à l'international.
rfm_hors_uk = rfm_avec_pays[rfm_avec_pays["Country"] != "United Kingdom"]
ca_hors_uk = rfm_hors_uk.groupby(["cluster", "Country"])["Montant"].sum()
top_pays_hors_uk = ca_hors_uk.groupby("cluster").idxmax().apply(lambda x: x[1])
tableau_segments["pays_dominant_hors_UK"] = tableau_segments["cluster"].map(top_pays_hors_uk)
tableau_segments
```

```python
# Nommage automatique des clusters à partir de leur profil moyen, classés
# du plus au moins engagé (Montant moyen décroissant). Ordre à vérifier
# et ajuster si besoin après lecture du tableau ci-dessus.
noms_ordonnes = ["Champions", "Clients réguliers", "Clients à risque", "Clients endormis"]
ordre_clusters = tableau_segments.sort_values("Montant_moyen", ascending=False)["cluster"].tolist()
correspondance_noms = dict(zip(ordre_clusters, noms_ordonnes[:len(ordre_clusters)]))

rfm["nom_segment"] = rfm["cluster"].map(correspondance_noms)
tableau_segments["nom_segment"] = tableau_segments["cluster"].map(correspondance_noms)
tableau_segments
```

## 11. Discussion critique et éthique (Partie 5 du TP)

`Impact des annulations` : mesuré ci-dessus par ARI  à commenter dans
le rapport avec la valeur obtenue sur les vraies données.

`Pourquoi log + standardisation plutôt que RFM bruts` : sans cette
étape, Montant (grande échelle, très asymétrique) domine le calcul de
distance de K-means, et les segments seraient surtout basés sur la
dépense, en ignorant Récence et Fréquence.

`Redondance Fréquence/Montant` : quantifiée par leur corrélation
(section 5)  à interpréter dans le rapport : est-elle assez forte pour
justifier une réduction de dimension, ou reste-t-elle raisonnable ?

`Limites de la silhouette seule`: elle favorise des clusters convexes
et bien séparés, sans lien avec leur utilité métier, et peut être
maximisée par un k trivial. D'où le croisement avec le coude, la
stabilité (ARI) et la comparaison à la baseline plus haut.

`Absence de vérité terrain` : la qualité des segments est défendue par
la stabilité (ARI ré-échantillonnage, sensibilité aux valeurs extrêmes),
la cohérence avec la baseline par règles, et le sens métier des profils
obtenus  pas par une métrique de justesse absolue, qui n'existe pas ici.

`Éthique de la tarification différenciée` : transparence sur les
critères utilisés, pas de proxy de caractéristiques protégées (origine,
âge, etc.) dans le choix des variables, plafond raisonnable d'écart de
prix entre segments, conformité RGPD si des données personnelles sont
exploitées, et droit du client à comprendre pourquoi il reçoit telle
offre plutôt qu'une autre.
